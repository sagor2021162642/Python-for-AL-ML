```python
def batch_generator(data_samples, batch_size):
    for i in range(0,len(data_samples),batch_size):
        yield data_samples [i:i+batch_size]
```

```python
data = [1, 2, 3, 4, 5, 6, 7, 8]
batch_size = 3

for batch in batch_generator(data,batch_size):
    print(batch)
```
