```python
n = input().split()
brightness  = float(n[0])
threshold = float(n[1])
#print(brightness)
#print(threshold)
if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```

```python
n = input().split()
x = float(n[0])
min_v = float(n[1])
max_v = float(n[2])
norm = (x - min_v) / (max_v - min_v)
print(f"{norm:.2f}")
```

```python
n = int(input())

target = float(input())

average_loss = 0
total_loss = 0

for i in range(n):
    loss = float(input())
    total_loss = total_loss + loss

average_loss = total_loss/n

if average_loss <= target:
    print("PASS")
else:
    print("RETRY")
```

```python
n = int(input())

yes_counter = 0
no_counter = 0


for i in range(n):
    vote = input().upper()
    if vote == "YES":
        yes_counter = yes_counter + 1
    else:
        no_counter = no_counter + 1
if yes_counter >= no_counter:
    print("ACCEPT")
else:
    print("REJECT")
```

```python
message = input().lower()
if "happy" in message:
    print("Happy Mood")
elif "joy" in message:
    print("Happy Mood")
elif"smile" in message:
    print("Happy Mood")
elif"sad" in message:
    print("Sad Mood")
elif"cry" in message:
    print("Sad Mood")
elif"angry" in message:
    print("Sad Mood")
else:
    print("Neutral Mood")
```

```python
pixel_brightness_values = input().split()

total_brightness = 0

for i in range(len(pixel_brightness_values)):
    total_brightness = total_brightness + int(pixel_brightness_values[i])

average_brightness = total_brightness/len(pixel_brightness_values)

if average_brightness < 85:
    print("Dark Image")
elif 85 <= average_brightness <= 170:
    print("Normal Image")
elif average_brightness > 170:
    print("Bright Image")
```

```python
number_of_predictions = input().upper().split()
count_A = 0
count_B= 0

for i in range(len(number_of_predictions)):
    if "A" in number_of_predictions[i]:
        count_A = count_A + 1
    else:
        count_B = count_B + 1
percentage_of_A =  float(count_A / len(number_of_predictions))
percentage_of_B =  float(count_B / len(number_of_predictions))

if percentage_of_A > 0.70:
    print("Biased Model")
elif percentage_of_B > 0.70:
    print("Biased Model")
else:
    print("Fair Model")
```

```python
user_message = input().lower().split()
keyword_counter = 0
for i in range(len(user_message)):
    if user_message[i] ==  "ai":
        keyword_counter = keyword_counter + 1
    elif user_message[i] ==  "data":
        keyword_counter = keyword_counter + 1
    elif user_message[i] ==  "model":
        keyword_counter = keyword_counter + 1
    elif user_message[i] ==  "learn":
        keyword_counter = keyword_counter + 1
    elif user_message[i] ==  "train":
        keyword_counter = keyword_counter + 1
    elif user_message[i] ==  "neural":
        keyword_counter = keyword_counter + 1
if keyword_counter >= 2:
    print("AI Detected")
else:
    print("Not AI Related")
```
