```python
n = int(input())
n_list = input().split()
new_n_list = []
for i in range(n):
    new_n_list.append(int(n_list[i]))

lowest = min(new_n_list)
l_position = new_n_list.index(lowest)
print(lowest, l_position+1)
```

```python
n = int(input())
n_list = input().split()
new_n_list = []
for i in range(n):
    new_n_list.append(int(n_list[i]))

minimum  = min(new_n_list)
maximum = max(new_n_list)
position_of_minimum = new_n_list.index(minimum)
position_of_maximum = new_n_list.index(maximum)

new_n_list[position_of_minimum] = maximum
new_n_list[position_of_maximum] = minimum

for j in range(n):
    print(new_n_list[j],end=" ")
```
